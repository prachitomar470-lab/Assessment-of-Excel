```python
import pandas as pd

df = pd.read_csv("Task4_Data.csv")

df.head()
```

```python
df.describe()
```

```python
Q1 = df["DeliveryTime"].quantile(0.25)
Q3 = df["DeliveryTime"].quantile(0.75)

IQR = Q3 - Q1

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
```

```python
correlation_matrix = df[["Revenue", "DeliveryTime", "Rating"]].corr()

correlation_matrix
```

```python
# Negative correlation between DeliveryTime and Revenue suggests that longer delivery times
# are associated with lower revenue in this sample.

# Negative correlation between DeliveryTime and Rating suggests that longer delivery times
# are associated with lower customer ratings in this sample.

# Positive correlation between Revenue and Rating suggests that higher revenue orders
# are associated with higher ratings in this sample.
```
